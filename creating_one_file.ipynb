```python
def load_p(filename):
    return pd.read_parquet(os.path.join(data_path, filename))



print("Loading files...")
amz_reviews = load_p('Books_rating.parquet')
amz_meta = load_p('amazon_books_metadata.parquet')
bx_meta = load_p('bx_books_metadata.parquet')
bx_ratings = load_p('bx_ratings.parquet')
bx_users = load_p('bx_users.parquet')
gr_books = load_p('goodreads_books_part_000.parquet')
print("All files loaded successfully!\n" + "="*50)
```

```python
# -------------------------------------------------------------------
# 1. PROCESS STREAM A: AMAZON
# -------------------------------------------------------------------
print("Starting with Amazon:")

display(amz_reviews.head())
display(amz_meta.head())
```

```python
print("\n--- Processing Amazon Stream ---")

# Standardize score scale 1-5 -> 0.0-1.0
amz_reviews['normalized_score'] = (amz_reviews['review/score'] - 1.0) / (5.0 - 1.0)

# Structure individual reviews as a dict
amz_reviews['review_dict'] = amz_reviews.apply(
    lambda row: {
        'user_id': str(row['User_id']),
        'original_score': row['review/score'],
        'normalized_score': row['normalized_score'],
        'summary': row['review/summary'],
        'text': row['review/text'],
        'user_demographics': None
    }, axis=1
)

display(amz_reviews.head())

```

```python
# Aggregate reviews per book Title
amz_grouped = amz_reviews.groupby('Id').agg(
    Title=('Title', 'first'),
    review_count=('review/score', 'count'),
    reviews_list=('review_dict', list)
).reset_index()

display(amz_grouped.head())
```

```python
# Join with Metadata on Title
amz_merged = pd.merge(amz_meta, amz_grouped, on='Title', how='left')
print(f"Amazon Total Books: {len(amz_merged):,}")
print(f"Amazon Books with >= 5 reviews: {(amz_merged['review_count'] >= 5).sum():,}")
```

```python
amz_5plus = amz_merged[amz_merged['review_count'] >= 5].copy()
display(amz_5plus.head())
print(len(amz_5plus))
```

```python

# -------------------------------------------------------------------
# 2. PROCESS STREAM B: BOOK-CROSSING
# -------------------------------------------------------------------
print("\n--- Processing Book-Crossing Stream ---")

display(bx_meta.head())
print(len(bx_meta))
display(bx_ratings.head())
print(len(bx_ratings))
display(bx_users.head())
print(len(bx_users))

```

```python
# Merge ratings with demographics
bx_reviews_full = pd.merge(bx_ratings, bx_users, on='User-ID', how='left')

display(bx_reviews_full)
```

```python
# Exclude implicit 0 ratings if desired (Optional: change to > 0 if you only want explicit ratings)
bx_explicit = bx_reviews_full[bx_reviews_full['Book-Rating'] > 0].copy()

# Clean ISBN text without adding or padding any characters.
# Keep digits and X, which is a valid ISBN-10 check character.
def clean_isbn(series):
    cleaned = (
        series.astype('string')
        .str.replace(r'[^0-9Xx]', '', regex=True)
        .str.upper()
    )
    return cleaned.mask(cleaned.eq('') | cleaned.str.fullmatch(r'0+'))

bx_explicit['ISBN'] = clean_isbn(bx_explicit['ISBN'])
bx_meta['ISBN'] = clean_isbn(bx_meta['ISBN'])

# Standardize score scale 1-10 -> 0.0-1.0
bx_explicit['normalized_score'] = (bx_explicit['Book-Rating'] - 1.0) / (10.0 - 1.0)

display(bx_explicit.head())
```

```python
# Structure individual review dict
bx_explicit['review_dict'] = bx_explicit.apply(
    lambda row: {
        'user_id': str(row['User-ID']),
        'original_score': row['Book-Rating'],
        'normalized_score': row['normalized_score'],
        'summary': None,
        'text': None,
        'user_demographics': {
            'age': row['Age'] if pd.notnull(row['Age']) else None,
            'location': row['Location'] if pd.notnull(row['Location']) else None
        }
    }, axis=1
)

display(bx_explicit.head())
```

```python
# Aggregate reviews per ISBN
bx_grouped = bx_explicit.groupby('ISBN').agg(
    review_count=('Book-Rating', 'count'),
    reviews_list=('review_dict', list)
).reset_index()

display(bx_grouped.head())
```

```python
# Join with Metadata on ISBN
bx_merged = pd.merge(bx_meta, bx_grouped, on='ISBN', how='left')
print(f"Book-Crossing Total Books: {len(bx_merged):,}")
print(f"Book-Crossing Books with >= 5 reviews: {(bx_merged['review_count'] >= 5).sum():,}")
```

```python
bx_5plus = bx_merged[bx_merged['review_count'] >= 5].copy()
display(bx_5plus.head())
print(len(bx_5plus))
```

```python

# -------------------------------------------------------------------
# 3. PROCESS STREAM C: GOODREADS
# -------------------------------------------------------------------
print("\n--- Goodreads Stream ---")
print(f"Goodreads Metadata Books: {len(gr_books):,}")
display(gr_books.head())
```

```python
gr_5plus = gr_books[gr_books['ratings_count'] >= 5].copy()
print(len(gr_5plus))
```

```python
# Standardize all three sources to one common schema.
# Reviews_List contains review dictionaries; Average_Normalized_Rating is always on a 0-1 scale.
common_columns = [
    'Title',
    'Clean_Title',
    'Description',
    'Authors',
    'Category/Genre',
    'ISBN/ID',
    'Book ID',
    'Review_Count',
    'Reviews_List',
    'Average_Normalized_Rating'
]


def average_review_score(reviews):
    """Return the mean normalized score from a list of review dictionaries."""
    if not isinstance(reviews, list):
        return np.nan

    scores = [
        review.get('normalized_score')
        for review in reviews
        if isinstance(review, dict) and pd.notna(review.get('normalized_score'))
    ]
    return np.mean(scores) if scores else np.nan


# Rebuild the filtered source dataframes so this cell can be rerun safely.
amz_source = amz_merged[amz_merged['review_count'] >= 5].copy()
bx_source = bx_merged[bx_merged['review_count'] >= 5].copy()
gr_source = gr_books[gr_books['ratings_count'] >= 5].copy()
goodreads_rating = pd.to_numeric(gr_source['average_rating'], errors='coerce')
goodreads_normalized_rating = ((goodreads_rating - 1.0) / 4.0).where(
    goodreads_rating.between(1.0, 5.0)
)

amz_standardized = pd.DataFrame({
    'Title': amz_source['Title'],
    'Clean_Title': amz_source['clean_title'],
    'Description': amz_source['description'],
    'Authors': amz_source['authors'],
    'Category/Genre': amz_source['categories'],
    'ISBN/ID': None,
    'Book ID': 'AMZ_' + amz_source['Id'].astype('string'),
    'Review_Count': amz_source['review_count'],
    'Reviews_List': amz_source['reviews_list'],
    'Average_Normalized_Rating': amz_source['reviews_list'].apply(average_review_score)
})

bx_standardized = pd.DataFrame({
    'Title': bx_source['Book-Title'],
    'Clean_Title': bx_source['clean_title'],
    'Description': None,
    'Authors': bx_source['Book-Author'],
    'Category/Genre': None,
    'ISBN/ID': bx_source['ISBN'],
    'Book ID': 'BX_' + pd.Series(range(len(bx_source)), index=bx_source.index).astype('string').str.zfill(7),
    'Review_Count': bx_source['review_count'],
    'Reviews_List': bx_source['reviews_list'],
    'Average_Normalized_Rating': bx_source['reviews_list'].apply(average_review_score)
})

# Goodreads average_rating is on a 1-5 scale; invalid values become missing.
gr_standardized = pd.DataFrame({
    'Title': gr_source['title'],
    'Clean_Title': gr_source['clean_title'],
    'Description': gr_source['description'],
    'Authors': None,
    'Category/Genre': gr_source['genres'],
    'ISBN/ID': gr_source['isbn'],
    'Book ID': 'GR_' + gr_source['book_id'].astype('string'),
    'Review_Count': gr_source['ratings_count'],
    'Reviews_List': None,
    'Average_Normalized_Rating': goodreads_normalized_rating
})

amz_5plus = amz_standardized[common_columns]
bx_5plus = bx_standardized[common_columns]
gr_5plus = gr_standardized[common_columns]

# The three dataframes now have identical columns and can be concatenated.
all_books = pd.concat([amz_5plus, bx_5plus, gr_5plus], ignore_index=True)

print('Common columns:', list(all_books.columns))
print(f'Amazon: {len(amz_5plus):,} | Book-Crossing: {len(bx_5plus):,} | Goodreads: {len(gr_5plus):,}')
print(f'Combined: {len(all_books):,}')
print('\nNormalized rating ranges:')
print(all_books['Average_Normalized_Rating'].describe()[['min', 'max', 'mean']])
display(all_books.head())
```

```python
# Save the combined dataset as a JSON array of book records.
# Write one record at a time to avoid loading the complete JSON document into memory.
import json

json_path = Path('all_books.json')
temp_json_path = json_path.with_suffix('.json.tmp')


def make_json_safe(value):
    if isinstance(value, dict):
        return {key: make_json_safe(item) for key, item in value.items()}
    if isinstance(value, (list, tuple)):
        return [make_json_safe(item) for item in value]
    if isinstance(value, np.generic):
        return make_json_safe(value.item())
    if value is None:
        return None

    try:
        if pd.isna(value):
            return None
    except (TypeError, ValueError):
        pass

    return value


with temp_json_path.open('w', encoding='utf-8') as output_file:
    output_file.write('[\n')

    for row_number, row in enumerate(all_books.itertuples(index=False, name=None)):
        if row_number:
            output_file.write(',\n')

        record = dict(zip(all_books.columns, row))
        json.dump(
            make_json_safe(record),
            output_file,
            ensure_ascii=False,
            separators=(',', ':')
        )

    output_file.write('\n]\n')

temp_json_path.replace(json_path)
print(f'Saved {len(all_books):,} records to {json_path.resolve()}')
```
